```python
import numpy as np
name=['Alice','Bob','Cathy','Dough']
age=[25,45,37,19]
weight=[55.0,85.5,68.0,61.5]
data=np.zeros(4,dtype={'names':('name','age','weight'),'formats':('U10','i4','f8')})
print("Structured array using a compound data type")
print(data.dtype)
data['name']=name
data['age']=age
data['weight']=weight
print("filled structured array")
print(data)
print("all names")
print(data['name'])
print("first row of data")
print(data[0])
print("name from the last row")
print(data[-1]['name'])
print("names where age is under 30")
print(data[data['age']<30]['name'])
np.dtype({'names':('name','age','weight'),'formats':('U10','i4','f8')})
np.dtype({'names':('name','age','weight'),'formats':((np.str_,10),int,np.float32)})
np.dtype([('name','S10'),('age','i4'),('weight','f8')])
np.dtype('S10','i4','f8')
```
