```python
import numpy as np
a=np.arange(1,21)
print(a)
print(a.ndim)
print(a.shape)
print(a.size)
print(a.dtype)
```

```python
np.random.seed(0)
x=np.random.randint(50,size=(4,5))
print(x)
print(x.nbytes)
print(x.itemsize)
```

```python
arr=np.arange(1,13).reshape(3,4)
print(arr[2,3])
print(arr[-1])
print(arr[:,-1])
```

```python
x=np.arange(1,13)
y=x.reshape(3,4)
z=x.reshape(12,1)
a=x[:,np.newaxis]
print(y)
print(z)
print(a)
```

```python
x=np.arange(1,13).reshape(3,4)
a=x[:2,:2]
b=x[:,::2]
c=x[:,::-1]
print(a)
print(b)
print(c)
```

```python
x=np.arange(1,21).reshape(4, 5)
a=x[:3,:4]
a[0,0]=100
print(a)
print(x)
b=x[:3,:4].copy()
b[0,0]=200
print(b)
print(x)
```

```python
a=np.random.randint(1,10,(2,3))
b=np.random.randint(1,10,(2,3))
print(a)
print(b)
c=np.vstack((a,b))
d=np.concatenate((a,b),axis=0)
print(c)
print(d)
e=np.hstack((a,b))
f=np.concatenate((a,b),axis=1)
print(e)
print(f)
```

```python
x=np.arange(1,16)
print(x)
a=np.split(x,3)
print(a[0])
print(a[1])
print(a[2])
```

```python
x=np.arange(1,25).reshape(6,4)
print(x)
a=np.hsplit(x,2)
print(a[0])
print(a[1])
b=np.vsplit(x,2)
print(b[0])
print(b[1])
```
