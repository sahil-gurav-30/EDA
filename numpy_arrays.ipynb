```python
import numpy as x
arr=x.array([10,20,30,40,50])
print(arr)
print(type(arr))
```

```python
b=x.array([[1,2,3],[4,5,6]])
print(b)
print(type(b))
```

```python
a=x.array([7,8,9],dtype=float)
print(a)
print(type(a))
```

```python
c=x.zeros(6)
d=x.ones((2,3),dtype=int)
print(c)
print(d)
```

```python
e=x.empty((2,2))
print(e)
```

```python
f=x.arange(0,20,2)
g=x.linspace(0,1,6)
print(f)
print(g)
```
