```python
import numpy as np

x = np.random.randint(1, 100, 10)
print("Original array:", x)

y = np.sort(x)
print("After np.sort():", y)
print("Original unchanged:", x)

x.sort()
print("After x.sort():", x)
```

```python
import numpy as np

x = np.random.randint(1, 100, 10)
print("Original array:", x)

i = np.argsort(x)
print("Sorted indices:", i)

print("Using fancy indexing:", x[i])
print("Using np.sort():", np.sort(x))

print("Both are equal:", np.array_equal(x[i], np.sort(x)))
```

```python
import numpy as np

x = np.random.randint(1, 100, (4, 6))
print("Original array:\n", x)

row_sorted = np.sort(x, axis=1)
print("\nRows sorted independently:\n", row_sorted)

col_sorted = np.sort(x, axis=0)
print("\nColumns sorted independently:\n", col_sorted)
```

```python
import numpy as np

x = np.random.randint(1, 100, 15)
print("Original array:", x)

p = np.partition(x, 4)
print("Partitioned array:", p)
print("Four smallest values:", p[:4])

print("Verified using np.sort():", np.sort(x)[:4])
```

```python
import numpy as np

student_names = np.array([
    'Rahul', 'Priya', 'Amit', 'Sneha', 'Kiran'
])

scores = np.array([85, 92, 76, 95, 88])

i = np.argsort(scores)

print("Students ranked from lowest to highest:")
for j in i:
    print(student_names[j], scores[j])
```

```python
import numpy as np

X = np.random.rand(10, 2)
print("Points:\n", X)

dist_sq = np.sum(
    (X[:, np.newaxis, :] - X[np.newaxis, :, :]) ** 2,
    axis=-1
)

print("\nPairwise squared-distance matrix:\n", dist_sq)

i = np.argsort(dist_sq, axis=1)

# Exclude each point itself (distance = 0)
nearest = i[:, 1:3]

print("\nIndices of 2 nearest neighbors:")
for j in range(10):
    print("Point", j, "->", nearest[j])
```
